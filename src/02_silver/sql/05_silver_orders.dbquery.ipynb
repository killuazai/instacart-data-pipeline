```sql
-- saved-query name: 05_silver_orders
-- TABLE: ORDERS_CLEAN
-- PURPOSE: Create a cleaned Silver version of the orders data. Standardize text values while preserving the original order data and meaningful null values.

USE CATALOG workspace;
CREATE SCHEMA IF NOT EXISTS instacart_silver;
USE SCHEMA instacart_silver;

CREATE OR REPLACE TABLE orders_clean AS
SELECT
    order_id,
    user_id,

    -- Standardize text formatting
    LOWER(TRIM(eval_set)) AS eval_set,

    -- Preserve numeric order fields
    order_number,
    order_dow,
    order_hour_of_day,

    -- Preserve NULL for first orders since there is no prior order
    days_since_prior_order

FROM workspace.instacart_bronze.orders_raw;
```
